```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

## 1. Load data

```python
df = pd.read_csv("../Data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
print(df.shape)
df.head()
```

## 2. Clean data

```python
df = df.drop_duplicates(subset="customerID")
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce").fillna(0)
df["ChurnFlag"] = (df["Churn"] == "Yes").astype(int)
df["TenureGroup"] = pd.cut(df["tenure"], bins=[-1, 3, 12, 24, 100],
                           labels=["0-3 months", "4-12 months", "1-2 years", "2+ years"])
print("Missing values:", df.isna().sum().sum())
```

## 3. KPIs

```python
total = len(df)
churned = int(df["ChurnFlag"].sum())
churn_rate = churned / total * 100
revenue_lost = df.loc[df["ChurnFlag"] == 1, "MonthlyCharges"].sum()
print(f"Total customers : {total}")
print(f"Churned         : {churned}")
print(f"Churn rate      : {churn_rate:.2f}%")
print(f"Retention rate  : {100 - churn_rate:.2f}%")
print(f"Revenue lost    : ${revenue_lost:,.2f} (monthly charges of churned customers)")
```

## 4. Churn by segment

```python
def churn_by(col):
    return (df.groupby(col, observed=True)["ChurnFlag"].mean() * 100).round(2)

for col in ["Contract", "TenureGroup", "PaymentMethod", "InternetService", "TechSupport"]:
    s = churn_by(col)
    print(s, "\n")
    ax = s.plot(kind="bar", figsize=(7, 4), color="#4c78a8")
    ax.set_ylabel("Churn rate (%)")
    ax.set_title(f"Churn rate by {col}")
    for i, v in enumerate(s):
        ax.text(i, v + 0.5, f"{v}%", ha="center")
    plt.xticks(rotation=20)
    plt.savefig(f"../images/churn_by_{col}.png", dpi=150, bbox_inches="tight")
    plt.show()
```
